```python
import pandas as pd
```

#Series --> 1 Dimensional data representation
#DataFrame --> 2 Dimensional data representation (table -->rows,columns)

```python
marks = [45,65,23,45]
d = pd.Series(marks)
print(d)
print(type(d))
print(d.dtype)
print(d.ndim)
print(d.shape)
print(d.size)
print(type(d.dtypes))
print(d.index)
```

```python
data = {'names':['AAA','PFS','JFS','DA','DS'],
'ids':[12,43,56,67]}
h = pd.Series(data)
print(h)
print(h.index)
print(h.dtype)
print(h['names'])
print(h['ids'])
print(h['names'][2])
print(h['ids'][:2])
print(h['ids'])
```

```python
data = 6
g = pd.Series(data)
print(g)
h = pd.Series(data,index =['rank'])
print(h)
print(h.dtype)
```

```python
name = "Codegnan"
g = pd.Series(name)
print(g)
g = pd.Series(name,index=['company'])
print(g)
```

```python
#DataFrame -->2 Dimensional representation (table format)
marks = [45,65,23,45]
h = pd.DataFrame(marks)
print(h) #default index and column as rangeindex
print(h.index);print(h.columns)
k = pd.DataFrame(marks,columns =['Marks']) #modifying the column
print(k)
print(k.shape)
print(k.size)
print(k.ndim)
```

```python
details = {'Products':['Mobiles','Laptops','Earphones','Headphones'],
'Sales':[20000,10000,21000,50000],
'Stock':[10,5,35,55],
}
h = pd.DataFrame(details)
print(h)
print(h.shape)
print(h.size)
print(h.ndim)
print(type(h))
print(h.dtypes) #returns dtype of each column
print(h.select_dtypes('int'))
```

```python
#to access columns
print(h.columns)
```

```python
h['Products']
#Group of columns
h[['Products','Sales']]
```

```python
h[['Products','Sales']];
h.Products
```

```python
#Access rowwise data ->index -->loc(),iloc()
h.index
```

```python
print(h.loc[0]) #locating index values (Selecting by labels)
print(h.loc[0]['Products'])
print(h.loc[0][['Sales','Stock']])
print(h.loc[[2,0]][['Products','Stock']])
print(h.loc[1,'Stock'])
```

```python
h
```

```python
#iloc() -->selecting by position
print(h.index) #As its default integer index both loc(),iloc()
print(h.iloc[0])
print(h.iloc[[2,3]])
print(h.iloc[[2,3]][['Stock','Products']])
```

```python
print(h.info()) #Complete information about dataframe
#Summary of the dataframe -->Descrptive Statitical info
print(h.describe())
print(h.describe(include='object')) #for object type -->count,unique
#top --> most repeating value,freq -->count of most repeating value
```

```python
#Change the datatype of column
h['Sales'] = h['Sales'].astype('float')
print(h['Sales'].info())
```
